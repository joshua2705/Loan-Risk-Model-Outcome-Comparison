```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Patch
from matplotlib.lines import Line2D

# Data
models = ["XGBoost", "Random Forest", "GAM+WoE",
          "Logistic", "Rule-Fit", "Decision Tree"]
accuracy = [0.7161, 0.7242, 0.7250, 0.7146, 0.7374, 0.7000]
logistic_level = 0.7146

black_box = {"XGBoost", "Random Forest"}
colors = ["steelblue" if m in black_box else "mediumseagreen" for m in models]

plt.figure(figsize=(8, 4))
x = np.arange(len(models))

# Thinner bars: width < 1.0
plt.bar(x, accuracy, width=0.4, color=colors, edgecolor="black")
plt.axhline(y=logistic_level, color="red", linestyle="--",
            linewidth=2, label="Logistic accuracy")

plt.xticks(x, models, rotation=45, ha="right")
plt.ylabel("Accuracy")
plt.ylim(0.68, 0.75)
plt.title("Model Accuracy: Black-Box vs Interpretable")

# Legend outside the plot (right side)
legend_patches = [
    Patch(facecolor="steelblue", edgecolor="black", label="Black-box models"),
    Patch(facecolor="mediumseagreen", edgecolor="black", label="Interpretable models"),
    Line2D([], [], color="red", linestyle="--", label="Logistic accuracy")
]

plt.legend(handles=legend_patches, bbox_to_anchor=(1.02, 1),
           loc="upper left", borderaxespad=0.)

plt.tight_layout()
plt.show()
```
